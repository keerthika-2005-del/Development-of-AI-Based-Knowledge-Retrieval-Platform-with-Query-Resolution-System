Reuse the existing SQLite database connection and employees table created in previous cells.

Write an SQL query using SELECT * FROM employees; to retrieve all employee records.

Execute the query and fetch all the records.

Display the results in a neat table using a pandas DataFrame with the column names: employee_id, employee_name, department_id, salary, and email.

Display a suitable message if the table contains no records.

```python
import sqlite3
import pandas as pd
import os

# --- Setup: Create a dummy database and table if they don't exist ---
# This part ensures the code is runnable even if previous cells haven't been executed
# In a real scenario, you would skip this if you're sure the DB/table exist.

db_name = 'employees.db'
table_name = 'employees'

# Remove existing db for a clean run if desired (optional)
# if os.path.exists(db_name):
#     os.remove(db_name)
#     print(f"Removed existing {db_name} for a clean start.")

conn = None # Initialize conn to None for finally block

try:
    conn = sqlite3.connect(db_name)
    cursor = conn.cursor()

    # Create the employees table if it doesn't exist
    cursor.execute(f"""
        CREATE TABLE IF NOT EXISTS {table_name} (
            employee_id INTEGER PRIMARY KEY,
            employee_name TEXT NOT NULL,
            department_id INTEGER,
            salary REAL,
            email TEXT
        );
    """)

    # Insert some sample data if the table is empty
    cursor.execute(f"SELECT COUNT(*) FROM {table_name};")
    if cursor.fetchone()[0] == 0:
        print(f"Inserting sample data into {table_name}...")
        cursor.execute(f"""
            INSERT INTO {table_name} (employee_id, employee_name, department_id, salary, email) VALUES
            (1, 'Alice Smith', 101, 75000.00, 'alice.s@example.com'),
            (2, 'Bob Johnson', 102, 80000.00, 'bob.j@example.com'),
            (3, 'Charlie Brown', 101, 60000.00, 'charlie.b@example.com'),
            (4, 'Diana Prince', 103, 95000.00, 'diana.p@example.com');
        """)
        conn.commit()
    else:
        print(f"{table_name} table already contains data.")

    print(f"Connected to {db_name} and ensured {table_name} table exists.")

    # --- End of Setup ---

    # SQL query to retrieve all records
    sql_query = f"SELECT * FROM {table_name};"

    # Execute the query and fetch all records
    cursor.execute(sql_query)
    records = cursor.fetchall()

    # Define the column names as requested
    column_names = ['employee_id', 'employee_name', 'department_id', 'salary', 'email']

    # Display the results in a pandas DataFrame
    if records:
        df = pd.DataFrame(records, columns=column_names)
        print("\nEmployee Records:")
        print(df)
    else:
        print("\nNo records found in the employees table.")

except sqlite3.Error as e:
    print(f"An SQLite error occurred: {e}")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
finally:
    # Close the database connection
    if conn:
        conn.close()
        print(f"\nDatabase connection to {db_name} closed.")
```

Import the sqlite3 and pandas libraries.

Reuse the existing SQLite database connection and employees table created in previous cells.

Write an SQL query to display employees whose salary is greater than 50,000 using the WHERE clause.

Use the SQL query: SELECT * FROM employees WHERE salary > 50000;

Execute the query and display the matching employee records in a neat pandas DataFrame.

Display a suitable message if no employees match the condition.

```python
import sqlite3
import pandas as pd
import os # For cleaning up the database file if it exists

# --- Setup: Simulate existing database and table (as requested) ---
# This part ensures the script is runnable even if previous cells haven't been executed.
# In a real scenario, 'employees.db' would already exist from prior operations.

DB_FILE = 'employees.db'

# Remove the database file if it already exists to ensure a clean slate for demonstration
# In a real "reuse" scenario, you wouldn't delete it.
if os.path.exists(DB_FILE):
    os.remove(DB_FILE)
    print(f"Removed existing database file: {DB_FILE}")

# Connect to the SQLite database (this will create the file if it doesn't exist)
conn = sqlite3.connect(DB_FILE)
cursor = conn.cursor()

# Create the employees table if it doesn't exist
cursor.execute('''
    CREATE TABLE IF NOT EXISTS employees (
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        name TEXT NOT NULL,
        position TEXT,
        salary INTEGER
    );
''')

# Insert some sample data, ensuring some records meet the condition and some don't.
# Using INSERT OR IGNORE to prevent duplicate insertions if the table already has data
employees_data = [
    ('Alice Smith', 'Software Engineer', 75000),
    ('Bob Johnson', 'Project Manager', 90000),
    ('Charlie Brown', 'Data Analyst', 48000), # Below 50k
    ('David Lee', 'UX Designer', 62000),
    ('Eve Davis', 'HR Specialist', 45000),   # Below 50k
    ('Frank Green', 'Marketing Director', 110000),
    ('Grace Hall', 'Intern', 35000),         # Below 50k
    ('Heidi Miller', 'Accountant', 50000)    # Exactly 50k, should NOT be included (salary > 50000)
]

cursor.executemany("INSERT OR IGNORE INTO employees (name, position, salary) VALUES (?, ?, ?)", employees_data)
conn.commit()
print("Database and 'employees' table initialized with sample data.")

# --- End of Setup ---


# 1. Define the SQL query
sql_query = "SELECT * FROM employees WHERE salary > 50000;"
print(f"\nExecuting SQL query: '{sql_query}'\n")

# 2. Execute the query using pandas and fetch results into a DataFrame
try:
    df = pd.read_sql_query(sql_query, conn)

    # 3. Display the results or a message if no records match
    if df.empty:
        print("No employees found with a salary greater than 50,000.")
    else:
        print("Employees with salary greater than 50,000:")
        print(df)

except pd.io.sql.DatabaseError as e:
    print(f"An error occurred while executing the SQL query: {e}")
except Exception as e:
    print(f"An unexpected error occurred: {e}")

finally:
    # 4. Close the database connection
    if conn:
        conn.close()
        print(f"\nDatabase connection to {DB_FILE} closed.")

```

Import the sqlite3 and pandas libraries.

Reuse the existing SQLite database connection and employees table created in previous cells.

Write an SQL query using the MAX() aggregate function to find the highest salary.

Use the SQL query: SELECT MAX(salary) AS highest_salary FROM employees;

Execute the query and display the highest salary clearly.

If the table is empty or contains no valid salary values, display a suitable message.

```python
import sqlite3
import pandas as pd
import os # Import os to remove the database file for a clean run if desired

# --- Simulate previous cells: Database connection and table creation/data insertion ---

# Define the database file name
db_file = 'employees.db'

# Remove the database file if it exists to ensure a clean run for demonstration
# You might want to comment this out if you truly want to reuse an existing, populated db
if os.path.exists(db_file):
    os.remove(db_file)
    print(f"Removed existing database file: {db_file}")

# Establish a connection to the SQLite database
# If the database file does not exist, it will be created.
conn = sqlite3.connect(db_file)
cursor = conn.cursor()

# Create the 'employees' table if it doesn't exist
# Using REAL for salary to allow for decimal values
cursor.execute('''
    CREATE TABLE IF NOT EXISTS employees (
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        name TEXT NOT NULL,
        salary REAL
    )
''')

# Insert some sample data into the employees table
# We'll include a NULL salary to test MAX() behavior
sample_data = [
    ('Alice', 60000.00),
    ('Bob', 75000.50),
    ('Charlie', 90000.75),
    ('David', 60000.00),
    ('Eve', 82000.25),
    ('Frank', None) # Employee with no recorded salary
]

# Clear existing data before inserting new, for consistent demonstration
cursor.execute("DELETE FROM employees")

cursor.executemany("INSERT INTO employees (name, salary) VALUES (?, ?)", sample_data)

# Commit the changes to the database
conn.commit()

print("Database and 'employees' table created and populated with sample data.")
print("-" * 30)

# --- End of simulated previous cells ---


# SQL query to find the highest salary using MAX()
sql_query = "SELECT MAX(salary) AS highest_salary FROM employees;"

# Execute the query
cursor.execute(sql_query)

# Fetch the result
# fetchone() returns a tuple (value,) or (None,) if no rows or all salaries are NULL
highest_salary_result = cursor.fetchone()

# Display the highest salary clearly
# MAX() returns NULL if there are no rows or if all values in the column are NULL
if highest_salary_result and highest_salary_result[0] is not None:
    print(f"The highest salary found is: ${highest_salary_result[0]:,.2f}")
else:
    print("No valid salary found or the employees table is empty.")

# --- Test case for an empty table or table with only NULL salaries ---
print("\n" + "=" * 30)
print("Testing with an empty table scenario:")

# Clear the employees table
cursor.execute("DELETE FROM employees")
conn.commit()
print("Employees table has been cleared.")

# Re-execute the query for the empty table
cursor.execute(sql_query)
highest_salary_empty_table = cursor.fetchone()

if highest_salary_empty_table and highest_salary_empty_table[0] is not None:
    print(f"The highest salary found is: ${highest_salary_empty_table[0]:,.2f}")
else:
    print("No valid salary found or the employees table is empty.")

print("\n" + "=" * 30)
print("Testing with only NULL salaries scenario:")

# Insert only NULL salaries
cursor.execute("DELETE FROM employees")
cursor.execute("INSERT INTO employees (name, salary) VALUES ('Grace', NULL)")
cursor.execute("INSERT INTO employees (name, salary) VALUES ('Heidi', NULL)")
conn.commit()
print("Employees table populated with only NULL salaries.")

# Re-execute the query for the NULL salaries table
cursor.execute(sql_query)
highest_salary_null_table = cursor.fetchone()

if highest_salary_null_table and highest_salary_null_table[0] is not None:
    print(f"The highest salary found is: ${highest_salary_null_table[0]:,.2f}")
else:
    print("No valid salary found or the employees table is empty.")


# Close the database connection
conn.close()
print("\nDatabase connection closed.")
```

Write an SQL query using GROUP BY and COUNT() to count employees in each department.

Use the SQL query:

SELECT department_id, COUNT(*) AS employee_count FROM employees GROUP BY department_id;

Execute the query and display the results in a neat pandas DataFrame.

Display each department ID along with its total number of employees

```python
import sqlite3
import pandas as pd

# --- Database Setup (simulating previous cells) ---
# This part ensures we have a database and table to work with.
# If you already have a 'test.db' and 'employees' table from previous runs,
# this code will connect to it. If not, it will create them.

db_file = 'test.db'
conn = None # Initialize conn outside try-except for later use

try:
    conn = sqlite3.connect(db_file)
    cursor = conn.cursor()

    # Create the employees table if it doesn't exist
    cursor.execute('''
        CREATE TABLE IF NOT EXISTS employees (
            employee_id INTEGER PRIMARY KEY,
            first_name TEXT,
            last_name TEXT,
            department_id INTEGER,
            salary REAL
        )
    ''')

    # Insert some sample data if the table is empty
    cursor.execute("SELECT COUNT(*) FROM employees")
    if cursor.fetchone()[0] == 0:
        employees_data = [
            (1, 'Alice', 'Smith', 101, 60000),
            (2, 'Bob', 'Johnson', 102, 75000),
            (3, 'Charlie', 'Brown', 101, 62000),
            (4, 'Diana', 'Prince', 103, 80000),
            (5, 'Eve', 'Adams', 102, 78000),
            (6, 'Frank', 'White', 101, 65000),
            (7, 'Grace', 'Hopper', 103, 90000),
            (8, 'Heidi', 'Klum', 102, 70000)
        ]
        cursor.executemany("INSERT INTO employees VALUES (?, ?, ?, ?, ?)", employees_data)
        conn.commit()

    print(f"Connected to {db_file} and ensured 'employees' table exists with data.")

except sqlite3.Error as e:
    print(f"Database error: {e}")
    if conn:
        conn.close()
    exit() # Exit if database connection fails

# --- SQL Query and Pandas Display ---

# The SQL query to count employees per department
sql_query = "SELECT department_id, COUNT(*) AS employee_count FROM employees GROUP BY department_id;"

print(f"\nExecuting SQL query:\n{sql_query}\n")

# Execute the query and load results directly into a pandas DataFrame
try:
    df_employee_counts = pd.read_sql_query(sql_query, conn)

    # Display the results
    print("Employee count per department:")
    print(df_employee_counts)

except pd.io.sql.DatabaseError as e:
    print(f"Error executing query or loading data into DataFrame: {e}")

finally:
    # It's good practice to close the connection when you're done
    if conn:
        conn.close()
        print("\nDatabase connection closed.")
```
